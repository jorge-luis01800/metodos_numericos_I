#Metodo de biseccion

Queremos resolver la ecuacion $x^3+4x-10=0$ en el intervalo $[1,2]$.
Para aplicar el **metodo de biseccion** verificamos el signo de $f(x)$ en los extremos del intervalo $[a_n,b_n]\in[1,2]$
De modo que $f(a_n)f(b_n)<0$.
Siendo $a_n=P_{n-1}$ si $f(P_{n-1})<0$ o $b_n=P_{n-1}$ si $f(P_{n-1})>0$.
Aproximaremos la solucion calculando de manera iterativa
$$P_n=\frac{b_n-a_n}{2}$$

Importamos la libreria numpy, que tiene funciones matematicas y matplotlib, que permite graficas

```python
#from math import *
import numpy as np
import matplotlib. pyplot as plt
```

Definimos la funcion $f(x)$ asociada a la ecuacion $f(x)=0$

```python
def f(x):
  return x**3 + 4*x**2 -10.0
```

Graficamente

```python
x=np.linspace(0,3,50) #geneta una reticula o una participn
plt.plot(x,f(x))
plt.title("Grafica $f(x)=x^3+4x-10$")
plt.xlabel("$x$")
plt.ylabel("$f(x)$")
plt.axhline(y=0, color='red') #trazar linea horizontal
plt.scatter(1.36,0, color='pink', label='Raiz')
plt.legend()
```

Estos son los hiperparametros y criterios de terminacion del algoritmo
