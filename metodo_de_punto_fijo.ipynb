#Metodo del punto fijo

Dadas las definiciones:\
Sean $f,g\in C[a,b]$ con $p\in[a,b]$\
$p$ es raiz de $f$ si $f(p)=0$\
$p$ es punto fijo de $g$ si $g(p)=p$ <font color="cyan">**...(*)**</font>\
de modo que dado el problema de encontrar $p\in[a,b]$, talque\
$f(p)=0$\
es equivalente si definimos\
$g(x)=x-f(x)$\
Para econtrar $p\in[a,b]$, talque\
$g(p)=p-f(p)=p-0=p$\
Es decir, solo cambiamos la forma en que vemos a la raiz de la funcion.\
Hacemos esto pues existe un metodo capaz de econtrar un punto fijo de una funcion.


##¿Como encontramos un punto fijo?
Particularmente, teniendo $g(x)\in C[a,b]$\
La podemos llevarla a una expresion de punto fijo:\
$x=g_i(x)$ (con infinitas posibilidades)\
De modo que para seleccionar la mas optima, consideremos el Teorema:

Sea $g(x)\in C[a,b]$ tal que $g(x)\in [a,b]$ $\forall x\in [a,b]$\
Suponga que existe $g´(x)\in(a,b)$ y existe una constante $0<k<1$ talque:\
$$|g'(x)|\leq k, \forall x\in(a,b)$$
Entonces, para cualquier $p_0\in[a,b]$ la sucesion definida por:
$$p_n = g(p_n-1), n\geq 1$$
Converge al unico punto fijo $p\in[a,b]$

Es decir para conocer $p$, necesitamos:


*   $g(x)$ reorganizada y/o ajustada como $x=g_i(x)$, (pues buscaremos $x\in[a,b]$ que cumpla la igualdad)
*   $g_i(x)\in [a,b]$ $\forall x\in [a,b]$
*   $g_i(x)$ continua en $[a,b]$
*   $g_i´(x)$ exista en $(a,b)$
*   Una costante $0<k<1$ tal que $|g'(x)|\leq k, \forall x\in(a,b)$



Consideremos la ecuacion propuesta (de hecho continua):\
$$f(x)=x^3 + 4x^2 -10$$
Para el intervalo $[1,2]$

Si $f(x)=0$\
$\implies x^3 + 4x^2 -10 =0$\
$\implies x^3 + 4x^2=10$\
$\implies x^2(x + 4)=10$\
$\implies x^2=\frac{10}{x + 4}$\
$\implies x=(\frac{10}{x + 4})^\frac{1}{2}$\
Tomamos la parte positiva de la raiz, pues $x\in[1,2]$



Sea $g(x)=(\frac{10}{x + 4})^\frac{1}{2}$\
Derivando:\
$g'(x)=-\frac{\sqrt{10}}{2(x+4)^{3/2}}$\
Para econtrar el maximo y minimo hagamos\
$-\frac{\sqrt{10}}{2(x+4)^{3/2}}=0$\
Como el numerador es distinto de cero y el denominador es positivo para todo $x\in[1,2]$, la ecuación no tiene soluciones\
Es decir, no existen puntos críticos en $(1,2)$, por lo que nos quedan como candidatos los extremos del intervalo


Evaluemos en $g(x)$

```python
def g(x):
    return (10/(x+4))**(1/2)
```

```python
print(f"g(1)={g(1)}")
```

```python
print(f"g(2)={g(2)}")
```

Asi, el máximo absoluto se alcanza en $x=1$\
Y el mínimo absoluto se alcanza en $x=2$\
Dado que:\
$1.2909944487358056\leq g(x)\leq 1.4142135623730951$ $\forall x\in[1,2]$\
Se cumple\
$$g(x)\in [a,b] \forall x\in [a,b]$$


Ahora para acotar la derivada\
Sea\
$h(x)=-\frac{\sqrt{10}}{2(x+4)^{3/2}}$\
Luego:\
$h'(x)=\frac{3\sqrt{10}}{4(x+4)^{5/2}}$\
Igualando a cero para econtrar los candidatos\
$\frac{3\sqrt{10}}{4(x+4)^{5/2}}=0$\
Nuevamente como el numerador es distinto de cero y el denominador es positivo para todo $x\in[1,2]$, la ecuación no tiene soluciones y solo nos quedan como candidatos los extremos:\
$h(1)=-\frac{\sqrt{10}}{2(5)^{3/2}}=−0.1414213562373095$\
$h(2)=-\frac{\sqrt{10}}{2(6)^{3/2}}=−0.1075828707279838$\
Asi\
$−0.1414213562373095\leq h(x)\leq −0.1075828707279838$ $\forall x\in[1,2]$\
Dado que $h(x)=g´(x)$\
$−0.1414213562373095\leq g'(x)\leq −0.1075828707279838$ $\forall x\in[1,2]$\
entonces\
$0.1414213562373095\geq -g'(x)\geq 0.1075828707279838$ $\forall x\in[1,2]$\
En particular como $g'(x)<0$, entonces $|g'(x)|=-g'(x)$, asi\
$0.1075828707279838\leq |g'(x)|\leq 0.1414213562373095$ $\forall x\in[1,2]$\
Y se cumple que:
$|g'(x)|\leq 0.1414213562373095 <1$

Por lo tanto, nuestra serie $p_n = g(p_n-1), n\geq 1$ va a converger a p


Traemos algunas bibloitecas de funciones y constantes:

```python
import numpy as np
import matplotlib.pyplot as plt
```

Definimos algunos valores necesarios para nuestro programa

```python
p0= 4 #valor inicial CUALQUIERA
TOL=0.000001 #Margen de precision
N_0=130      #Cantidad de iteaciones perimitidas
```

```python
#Inciamos el conteo de iteraciones en 1
i=1
#Imprimimos la parte inicial de la tabla
print(f"n iter\t\t P_n \t\t\t g(P_n) \t\t |P_n - P_n-1|")
print("{0} \t\t {1:6.15f} \t {2:6.15f} \t {3:6.15f}".format(i-1, p0, g(p0), abs(p0-g(p0))))
#Iniciamos un ciclo que finalice de acuerdo a la cantidad de iteraciones perimitida
while i<=N_0:
  p=g(p0) #Conozcamos el valor siguiente de la serie evaluando el p_i pasado
  if abs(p-p0)<TOL: #Analizamos la situacion segun lo calculado de p el p_n (¿terminamos o seguimos?)
    raiz=p          #Se ha cumplido, guardamos el punto fijo
    print("")
    print(f"La raiz aproximada de f es: {raiz}")
    break           #Terminamos el ciclo
  else:
    error= abs(p-p0) #Anter de actualizar, guardamos el error para imprimir
    i=i+1  #Avanzamos una interacion mas de la permitida
    p0=p   #El p calculado lo hacemos ahora el anteior para la serie
  print("{0} \t\t {1:6.15f} \t {2:6.15f} \t {3:6.15f}".format(i-1, p0, g(p0), error))
else:
  #En caso de no haber comprovado el TEOREMA, el metodo fallo
  print(f"El metodo fracaso despues de {N_0} iteraciones")
```
