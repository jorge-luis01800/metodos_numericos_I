```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def f(x):
  return 3*x**2 -np.exp(x)
```

```python
xx=np.linspace(0,1,50)
plt.plot(xx,f(xx))
plt.axhline(y=0, color='black')
```

```python
def newton(p0,tol):
  error=100
  Lerror=[]
  Lp_i=[p0]
  while error>tol:
    p1=p0-(3*p0**2-np.exp(p0))/(6*p0-np.exp(p0))
    error=abs(p0-p1)
    Lerror.append(error)
    Lp_i.append(p1)
    p0=p1
  return Lp_i,Lerror

```

```python
pd.DataFrame(newton(0.5,0.000001)).T
```
