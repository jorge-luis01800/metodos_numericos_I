#Definicion: Epsilon de la maquina

```python
def e(n):
  a=(n)/2
  return a, n

epsilon=1.0
for i in range(1700):
  epsilon, anterior=e(epsilon)
  if epsilon==0:
    print(f"El numero mas pequeño esta dado por: {anterior}")
    break
```
